```python
#jocul X si 0

import numpy as np

def afis_tabla(tabla):
  for i in range (3):
    a = tabla[3*i]
    b = tabla[3*i + 1]
    c = tabla[3*i + 2]
    print(f" {a} | {b} | {c}")

    if i < 2:
      print("-" * 9)

def pozitii_libere(tabla):
  return[i for i, celula in enumerate (tabla) if celula == ""]

def castigator(tabla,jucator):
  variante_de_castig = [
      (0,1,2), (3,4,5), (6,7,8),    #linii
      (0,3,6), (1,4,7), (2,5,8),    #coloane
      (0,4,8), (2,4,6)              #diagonale
  ]

  for a,b,c in variante_de_castig:
    if tabla[a] == jucator and tabla[b] == jucator and tabla[c] == jucator:
      return True

  return False

def remiza(tabla):

  if "" in tabla:
    return False
  if castigator(tabla, "X"):
    return False
  if castigator(tabla, "0"):
    return False

  return True

def minimax(tabla, jucator):
  """
  Scoruri: 1 daca X castiga, -1 daca 0 castiga, 0 - remiza
  Pp ca 'X' maximizeaza, '0' minimizeaza.
  """

  if castigator(tabla,"X"):
    return 1, None
  if castigator(tabla,"0"):
    return -1, None
  if remiza(tabla):
    return 0, None

  mutari = pozitii_libere(tabla)
  if jucator == "X":
    scor_maxim = -999
    mutare_optima = None
    for m in mutari:
      tabla_noua = tabla.copy()
      tabla_noua[m] = jucator
      scor, _ = minimax (tabla_noua, "0")

      if scor > scor_maxim:
        scor_maxim = scor
        mutare_optima = m
    return scor_maxim, mutare_optima
  else:     #jucator == "0" -> minimanizam
    scor_maxim = 999
    mutare_optima = None
    for m in mutari:
      tabla_noua = tabla.copy()
      tabla_noua[m] = jucator
      scor, _ = minimax(tabla_noua, "X")

      if scor < scor_maxim:
        scor_maxim = scor
        mutare_optima = m
    return scor_maxim, mutare_optima

def mutare_jucator(tabla):
  """Citim mutarea jucatorului"""
  mutari = pozitii_libere(tabla)
  while True:
    optiune = input(f"Alege o pozitie (0-8). Mutari disponibile: {mutari}: ")
    if optiune.isdigit() == False:
      print("Eroare! Va rugam sa introduceti un numar intre 0 si 8")
    optiune = int(optiune)

    if optiune in mutari:
      tabla[optiune] = "X"
      return
    else:
      print ("Eroare! Pozitie invalida sau ocupata. Va rugam sa incercati din nou")

def mutare_computer(tabla):
  scor, mutare = minimax(tabla, "0")
  if mutare is not None:
    tabla[mutare]= "0"
    print(f"Calculatorul a ales pozitia{mutare}.")

tabla = [""]*9
while True:
  print("Tabla curenta:")
  afis_tabla(tabla)

  mutare_jucator(tabla)

  if castigator(tabla, "X"):
    afis_tabla(tabla)
    print ("Felicitari! Ati castigat (X).")
    break

  if remiza(tabla):
    afis_tabla(tabla)
    print("Remiza!")
    break

  mutare_computer(tabla)
  if castigator(tabla, "0"):
    afis_tabla(tabla)
    print("Calcolatorul a castigat (0).")
    break

  if remiza(tabla):
    afis_tabla(tabla)
    print("Remiza!")
    break
```
